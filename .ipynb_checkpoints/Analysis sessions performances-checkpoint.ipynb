```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

```python
perf = pd.Series([75,80,80,80,80,85,85,85,93,95], name='performance')
```

```python
plt.Figure(dpi=300, figsize=(3,3))
sns.set(style='white', context='talk')

performances = [75,80,80,80,80,85,85,85,93,95]


sns.barplot(
    data=performances,
    estimator=np.mean,
    errorbar='sd',
    color='lightgray',
    capsize=.03,
    errwidth=1.5,
    width=.1)

plt.ylim([50,100])
plt.xlim([-.1,.1])
# Remove x-tick labels
plt.xticks([])
plt.ylabel('Performance (%)')
sns.despine()

# Define the directory and filename
directory = '/Users/bgoncalves/Desktop/Sofia - PhD/SVG Figures'
filename = 'F1_A.svg'

# Save the figure as an SVG file
fig = plt.gcf()  # Get the current figure
fig.savefig(os.path.join(directory, filename), format='svg')

plt.show()
```
